```python
import numpy as np

np.random.seed(42)

data = {
    "Student_ID": range(1, 201),
    "Attendance_Percentage": np.random.randint(50, 101, 200),
    "Study_Hours_Per_Day": np.round(np.random.uniform(1, 8, 200), 1),
    "Mathematics_Marks": np.random.randint(30, 101, 200),
    "Science_Marks": np.random.randint(30, 101, 200),
    "English_Marks": np.random.randint(30, 101, 200),
    "Computer_Science_Marks": np.random.randint(30, 101, 200)
}

df = pd.DataFrame(data)

df.head()
```

```python
print("Total Students:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
df.to_csv("student_performance.csv", index=False)

print("Dataset saved successfully!")
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
df.describe()
```

```python
subjects = [
    "Mathematics_Marks",
    "Science_Marks",
    "English_Marks",
    "Computer_Science_Marks"
]

subject_average = df[subjects].mean()

print("Subject-wise Average Marks:")
print(subject_average)
```

```python
plt.figure(figsize=(9, 5))

subject_average.plot(kind="bar")

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=20)
plt.tight_layout()

plt.show()
```

```python
average_attendance = df["Attendance_Percentage"].mean()

print("Average Student Attendance:", round(average_attendance, 2), "%")
```

```python
plt.figure(figsize=(9, 5))

plt.hist(df["Attendance_Percentage"], bins=10, edgecolor="black")

plt.title("Student Attendance Distribution")
plt.xlabel("Attendance Percentage")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
df["Average_Marks"] = df[
    ["Mathematics_Marks", "Science_Marks",
     "English_Marks", "Computer_Science_Marks"]
].mean(axis=1)

plt.figure(figsize=(9, 5))

plt.scatter(
    df["Study_Hours_Per_Day"],
    df["Average_Marks"]
)

plt.title("Study Hours vs Average Marks")
plt.xlabel("Study Hours Per Day")
plt.ylabel("Average Marks")

plt.tight_layout()
plt.show()
```

```python
correlation = df["Study_Hours_Per_Day"].corr(
    df["Average_Marks"]
)

print("Study Hours vs Average Marks Correlation:",
      round(correlation, 3))
```

```python
attendance_correlation = df["Attendance_Percentage"].corr(
    df["Average_Marks"]
)

print("Attendance vs Average Marks Correlation:",
      round(attendance_correlation, 3))
```

```python
correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Student Performance Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
def categorize_performance(marks):
    if marks >= 75:
        return "Excellent"
    elif marks >= 50:
        return "Average"
    else:
        return "Needs Improvement"


df["Performance_Category"] = df["Average_Marks"].apply(
    categorize_performance
)

print("Student Performance Categories:")
print(df["Performance_Category"].value_counts())
```

```python
category_counts = df["Performance_Category"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Performance Category Distribution")
plt.show()
```

```python
df.to_csv("student_performance_final.csv", index=False)

print("Final dataset saved successfully!")
print("Total Students:", len(df))
print("Total Columns:", len(df.columns))
```
